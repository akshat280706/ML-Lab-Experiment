```python
import numpy as np

num1 = np.array([[8,16],
                 [9,17]])

num2 = np.array([[28,43],
                 [30,45]])

print("Matrix 1:\n", num1)
print("Matrix 2:\n", num2)

# dot
print("\nDot Product")
print(np.dot(num1,num2))

# matmul
print("\nMatrix Multiplication")
print(np.matmul(num1,num2))

# inner
print("\nInner Product")
print(np.inner(num1,num2))

# outer
print("\nOuter Product")
print(np.outer(num1,num2))

# cross
a = np.array([8,16,9])
b = np.array([28,43,30])

print("\nCross Product")
print(np.cross(a,b))

# inverse
print("\nInverse")
print(np.linalg.inv(num1))

# determinant
print("\nDeterminant")
print(np.linalg.det(num1))

# eigenvalues & eigenvectors
values, vectors = np.linalg.eig(num1)

print("\nEigenvalues")
print(values)

print("\nEigenvectors")
print(vectors)

# norm
print("\nNorm")
print(np.linalg.norm(num1))

# solve
B = np.array([10,11])

print("\nSolution")
print(np.linalg.solve(num1,B))
```

```python
from sklearn.datasets import load_iris
from sklearn.model_selection import (
    cross_val_score,
    GridSearchCV,
    RandomizedSearchCV,
    KFold,
    StratifiedKFold
)
from sklearn.tree import DecisionTreeClassifier
from scipy.stats import randint

iris = load_iris()

X = iris.data
y = iris.target

model = DecisionTreeClassifier()

# cross_val_score
scores = cross_val_score(model,X,y,cv=5)
print("Cross Validation Scores")
print(scores)
print("Average Accuracy:",scores.mean())

# GridSearchCV
params = {
    "max_depth":[2,3,4,5],
    "criterion":["gini","entropy"]
}

grid = GridSearchCV(model,params,cv=5)
grid.fit(X,y)

print("\nGrid Search")
print(grid.best_params_)
print(grid.best_score_)

# RandomizedSearchCV
random = RandomizedSearchCV(
    model,
    {
        "max_depth":randint(2,10),
        "min_samples_split":randint(2,8)
    },
    n_iter=5,
    cv=5,
    random_state=42
)

random.fit(X,y)

print("\nRandom Search")
print(random.best_params_)
print(random.best_score_)

# KFold
print("\nKFold")
kf = KFold(n_splits=5,shuffle=True,random_state=42)

for train,test in kf.split(X):
    print("Train:",train)
    print("Test :",test)

# StratifiedKFold
print("\nStratified KFold")

skf = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

for train,test in skf.split(X,y):
    print("Train:",train)
    print("Test :",test)
```
